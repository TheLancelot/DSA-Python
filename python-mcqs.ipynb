```python
import numpy as np

a=np.ones((3,1))
b=np.ones( (1,4)) 
c=a+b
print (c) 

#In Python, standard lists use the + operator to concatenate (glue them together). However, NumPy arrays use the + operator for element-wise addition.When you try to add two NumPy arrays with different shapes—like a columns vector and a row vector—NumPy uses a powerful feature called broadcasting. Instead of throwing an error or concatenating, NumPy automatically stretches the arrays behind the scenes so their shapes match, and then adds them element-by-element.

#Two dimensions are compatible for NumPy operations if they meet at least one of these two conditions, checking from right to left (starting from the last dimension):
# They are exactly equal
# One of them is exactly 1

# If neither condition is met, NumPy will throw a ValueError: operands could not be broadcast together.
```

```python
import numpy as np
a=np.eye(3,5).T #dimension of this is (5,3)
b=np.eye (3,5).reshape((5,3)) #dimension of this is (5,3)
print(np.sum(a*b)) 

# print(a)
# print(b)

#for reshaping, you read elements row by row from the original matrix and place them row by row into the new matrix when doing this by hand.4

#In standard mathematics and linear algebra, you cannot multiply two (5x3))matrices together. 
# However, in NumPy, the * operator does not perform traditional matrix multiplication. Instead, it performs element-wise multiplication.Because a and b have the exact same shape, NumPy simply pairs up the elements at identical positions and multiplies them together
#np.sum(a*b) then adds up all those products to give you a single scalar value.

# if a and b had different shapes, we would have to use broadcasting rules to determine if they can be combined element-wise.
#  If they are compatible, NumPy would stretch one or both arrays as needed to make their shapes match before performing the multiplication. If they are not compatible, NumPy would raise a ValueError.

#brodcasting is outer product
a1=np.array([10,20,30]).reshape((3,1)) #dimension of this is (3,1)
b1=np.array([1,2])
print(a1+b1) #dimension of this is (3,2)
print(a1*b1)

```
